## Блок 1 · Один нейрон вручную
**Что делаем:** нейрон = взвешенная сумма входов + порог. Подбираем веса руками.
**Зачем:** увидеть, что «внутри» простая арифметика.

```python
import numpy as np
игрок = np.array([7, 4])        # [сила, скорость] из 10
веса  = np.array([0.5, 0.5])    # насколько важен каждый признак
порог = 5
сумма = np.dot(игрок, веса)
решение = 1 if сумма >= порог else 0
print(f'сумма={сумма}, решение={решение}')
```

## Блок 2 · Первая сеть на Keras (две «луны»)
**Что ожидаем:** сеть решает нелинейную задачу; точность заметно выше 50%.

```python
from sklearn.datasets import make_moons
from tensorflow import keras
X, y = make_moons(n_samples=400, noise=0.2, random_state=42)
model = keras.Sequential([
    keras.Input(shape=(2,)),
    keras.layers.Dense(8, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid'),
])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
history = model.fit(X, y, epochs=50, verbose=0)
print(f'Точность: {history.history["accuracy"][-1]:.0%}')
```

## Блок 3 · Кривая обучения
**Что ожидаем:** ошибка падает, точность растёт по эпохам.

```python
import matplotlib.pyplot as plt
plt.plot(history.history['loss'], label='ошибка')
plt.plot(history.history['accuracy'], label='точность')
plt.xlabel('эпоха'); plt.legend(); plt.title('Как сеть учится'); plt.show()
```

## Блок 4 · Граница решения (для наглядности)

```python
import numpy as np
xx, yy = np.meshgrid(np.linspace(-1.5,2.5,200), np.linspace(-1,1.5,200))
grid = np.c_[xx.ravel(), yy.ravel()]
zz = model.predict(grid, verbose=0).reshape(xx.shape)
plt.contourf(xx, yy, zz, alpha=0.5, cmap='coolwarm')
plt.scatter(X[:,0], X[:,1], c=y, cmap='coolwarm', edgecolor='k', s=20)
plt.title('Что выучила сеть'); plt.show()
```

---
**Итог.** Нейрон — взвешенная сумма + активация. Сеть подкручивает веса по эпохам, пока ошибка не станет маленькой. Правила вручную не пишем.
