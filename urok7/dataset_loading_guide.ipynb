### Открываем файл в pandas

⚠️ Название файла должно быть **точно такое же**, как ты загрузил.  
Замени `my_data.csv` на имя своего файла.

```python
# Замени 'my_data.csv' на название своего файла
df = pd.read_csv('my_data.csv')

df.head()
```

### ⚠️ Если возникла ошибка

| Ошибка | Что делать |
|---|---|
| `FileNotFoundError` | Проверь имя файла буква-в-букву |
| `UnicodeDecodeError` | Добавь `encoding='cp1251'` (см. ниже) |
| Данные в одном столбце | Добавь `sep=';'` (см. ниже) |

```python
# Пример: если файл в кодировке Windows
# df = pd.read_csv('my_data.csv', encoding='cp1251')

# Пример: если данные разделены точкой с запятой
# df = pd.read_csv('my_data.csv', sep=';')

# Пример: и то, и другое сразу
# df = pd.read_csv('my_data.csv', encoding='cp1251', sep=';')
```

# 🌐 Способ 2. Загрузка по URL (самый быстрый)

**Когда использовать:** датасет уже есть в интернете (GitHub, открытые источники).

### Пример 1. Датасет Titanic (классика ML)

```python
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'

df = pd.read_csv(url)
df.head()
```

### Пример 2. Датасет Pokemon

```python
url = 'https://gist.githubusercontent.com/armgilles/194bcff35001e7eb53a2a8b441e8b2c6/raw/92200bc0a673d5ce2110aaad4544ed6c4010f687/pokemon.csv'

df_pokemon = pd.read_csv(url)
df_pokemon.head()
```

### Пример 3. Датасет Iris (цветы)

```python
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv'

df_iris = pd.read_csv(url)
df_iris.head()
```

### ⚠️ Как правильно взять ссылку с GitHub

**❌ Так НЕ работает** (это страница, а не файл):  
`https://github.com/user/repo/blob/main/data.csv`

**✅ Так работает** (сырой файл):  
`https://raw.githubusercontent.com/user/repo/main/data.csv`

**Правило:** на GitHub нажми кнопку **«Raw»** → скопируй ссылку из адресной строки.
