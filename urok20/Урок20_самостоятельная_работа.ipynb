## Шаг 1 · Напиши функцию-предсказание
Сначала — обычная Python-функция. Проверь, что работает (без Gradio).

````python
def sentiment(text):
    positive=['отлично','супер','советую','нравится','класс']
    negative=['ужасно','плохо','долго','грубо']
    t=text.lower()
    p=sum(w in t for w in positive); n=sum(w in t for w in negative)
    return 'позитив' if p>n else 'негатив' if n>p else 'нейтрально'

print(sentiment('отличный сервис советую'))
print(sentiment('ужасно и долго'))
````

    positive=['отлично','супер','советую','нравится','класс']
