## Блок 1 (в Colab) · Gradio-интерфейс, временная ссылка
Для проверки: `share=True` даёт временную ссылку (гаснет с закрытием Colab).

````python
!pip install gradio -q
import gradio as gr
def sentiment(text):
    positive=['отлично','супер','советую','нравится']
    return 'позитив' if any(w in text.lower() for w in positive) else 'негатив'
gr.Interface(fn=sentiment, inputs=gr.Textbox(label='Отзыв'),
             outputs=gr.Label(label='Тональность'),
             title='Анализатор отзывов').launch(share=True)
````

## Блок 2 · Файл `app.py` (для Spaces)
Создай файл с именем **app.py** — Spaces сам его запустит:
```python
import gradio as gr

def sentiment(text):
    positive = ['отлично','супер','советую','нравится']
    return 'позитив' if any(w in text.lower() for w in positive) else 'негатив'

gr.Interface(fn=sentiment,
             inputs=gr.Textbox(label='Отзыв'),
             outputs=gr.Label(label='Тональность'),
             title='Анализатор отзывов').launch()
```

## Блок 3 · Файл `requirements.txt`
Список библиотек, которые Spaces установит:
```
gradio
```
(добавь `scikit-learn`, `tensorflow`, `transformers` — если модель их использует)

## Блок 4 · Публикация на Hugging Face Spaces
1. Зайти на **huggingface.co**, создать бесплатный аккаунт.
2. **New Space** → SDK: **Gradio**.
3. Загрузить **app.py** и **requirements.txt** (и файл модели, если есть).
4. Space соберётся автоматически и запустится.
5. Открыть постоянную ссылку `huggingface.co/spaces/<логин>/<имя>` и поделиться.
