Регрессия предсказывает **число** (не класс). Мы предскажем: сколько баллов наберёт ученик по числу часов подготовки.

## Шаг 1. Подключаем инструменты

```python
import pandas as pd
import numpy as np

print("Готово!")
```

## Шаг 2. Создаём данные: часы → баллы
Сделаем небольшую таблицу: сколько часов готовился и сколько баллов получил.

```python
np.random.seed(42)   # чтобы у всех вышло одинаково

hours = np.arange(1, 21)                              # часы: 1, 2, ... 20
scores = hours * 4 + 20 + np.random.normal(0, 5, 20)  # баллы (примерно 4 балла за час)

df = pd.DataFrame({"hours": hours, "score": scores.round(1)})  # собираем таблицу
df.head()
```

## Шаг 3. Смотрим на данные глазами
Построим точки — видно ли связь «больше часов → больше баллов»?

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.scatter(df["hours"], df["score"])   # точки: по X часы, по Y баллы
plt.xlabel("Часы подготовки")
plt.ylabel("Баллы")
plt.title("Связь часов и баллов")
plt.show()
```

## Шаг 4. Обучаем регрессию 📈
Модель проведёт через точки самую подходящую **прямую линию** — как линейку.

```python
from sklearn.linear_model import LinearRegression

X = df[["hours"]]   # признак — часы
y = df["score"]     # цель — баллы (ЧИСЛО)

model = LinearRegression()   # создаём модель
model.fit(X, y)              # обучаем — модель находит лучшую линию

print("Модель обучена!")
print("Наклон:", round(model.coef_[0], 1), "— столько баллов даёт каждый час")
```

## Шаг 5. Рисуем линию поверх точек

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["hours"], df["score"], label="реальные баллы")           # точки
plt.plot(df["hours"], model.predict(X), color="red", linewidth=2, label="линия")  # линия
plt.xlabel("Часы подготовки")
plt.ylabel("Баллы")
plt.legend()
plt.show()
```

## Шаг 6. Предсказываем 🔮
Сколько баллов наберёт тот, кто готовился 10 часов?

```python
new = pd.DataFrame([[10]], columns=["hours"])   # 10 часов как таблица

pred = model.predict(new)                       # модель предсказывает
print("Прогноз для 10 часов:", round(pred[0], 1), "баллов")
```

## Шаг 7. Насколько модель ошибается (MAE)
MAE = на сколько баллов в среднем модель промахивается. Меньше — лучше.

```python
from sklearn.metrics import mean_absolute_error

predictions = model.predict(X)                  # предсказания для всех
mae = mean_absolute_error(y, predictions)       # средняя ошибка
print("MAE:", round(mae, 1), "баллов")
```
