# Урок 8 — Регрессия: предсказываем ЧИСЛА

До сих пор мы предсказывали класс: выжил/не выжил, спорт/игры. Это **классификация**.

Сегодня новый тип задачи: предсказать **число**. Например, сколько баллов наберёт ученик,
который готовился 10 часов? Это **регрессия**.

Запускай ячейки по порядку: ▶ или **Shift+Enter**.

---
### Классификация vs Регрессия
- **Классификация** → ответ КЛАСС (кот/пёс, выжил/нет)
- **Регрессия** → ответ ЧИСЛО (баллы, цена, температура)

## Шаг 1. Создаём данные: часы подготовки → баллы
Сделаем небольшой понятный датасет: сколько часов готовился и сколько баллов получил.

```python
import pandas as pd
import numpy as np

np.random.seed(42)   # чтобы у всех получилось одинаково

hours = np.arange(1, 21)                        # 1..20 часов
scores = hours * 4 + 20 + np.random.normal(0, 5, 20)  # ~линейная связь + разброс

df = pd.DataFrame({"hours": hours, "score": scores.round(1)})
df.head()
```

## Шаг 2. Смотрим на данные глазами
Построим точки: видно ли связь «больше часов → больше баллов»?

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.scatter(df["hours"], df["score"])
plt.xlabel("Часов подготовки")
plt.ylabel("Баллы за экзамен")
plt.title("Больше готовишься — больше баллов?")
plt.show()
```

**Вопрос:** видишь, что точки идут вверх? Значит, связь есть — можно провести линию.

## Шаг 3. Делим на X и y, обучаем регрессию 📈
Линейная регрессия проводит через точки самую подходящую **прямую линию** — как линейка.

```python
from sklearn.linear_model import LinearRegression

X = df[["hours"]]   # признак — часы
y = df["score"]     # цель — баллы (ЧИСЛО!)

model = LinearRegression()
model.fit(X, y)   # модель находит лучшую прямую

print("Модель обучена!")
print("Наклон линии:", round(model.coef_[0], 1), "— столько баллов даёт каждый час")
```

## Шаг 4. Рисуем линию поверх точек
Посмотрим, как линия прошла через данные.

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["hours"], df["score"], label="реальные баллы")
plt.plot(df["hours"], model.predict(X), color="red", linewidth=2, label="линия регрессии")
plt.xlabel("Часов подготовки")
plt.ylabel("Баллы")
plt.legend()
plt.title("Линия, предсказывающая баллы")
plt.show()
```

## Шаг 5. Предсказываем 🔮
Сколько баллов наберёт тот, кто готовился 10 часов?

```python
# передаём число часов как таблицу
new = pd.DataFrame([[10]], columns=["hours"])

predicted = model.predict(new)
print("Прогноз для 10 часов подготовки:", round(predicted[0], 1), "баллов")
```

## Шаг 6. Насколько модель ошибается? (MAE)
Точки не лежат ровно на линии — модель ошибается. **MAE** говорит,
на сколько баллов в среднем модель промахивается.

```python
from sklearn.metrics import mean_absolute_error

predictions = model.predict(X)
mae = mean_absolute_error(y, predictions)
print("MAE:", round(mae, 1), "баллов")
print("Значит, в среднем модель ошибается примерно на", round(mae), "баллов.")
```
